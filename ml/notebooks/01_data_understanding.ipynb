## Column classification
| Type | Columns |
|---|---|

| **Numerical** | `amount (INR)`, `hour_of_day` |
| **Categorical** | `transaction type`, `merchant_category`, `transaction_status`, `sender_age_group`, `receiver_age_group`, `sender_state`, `sender_bank`, `receiver_bank`, `device_type`, `network_type`, `day_of_week` |
| **Binary** | `is_weekend` |
| **Date/time** | `timestamp` |
| **Irrelevant (ID)** | `transaction id` |

## Summary
- 250,000 rows, 17 columns, covering Jan 1 to Dec 30, 2024.
- Missing values: **0** in every column.
- Duplicate rows: **0**. Duplicate transaction ids: **0**.
- Target: `fraud_flag` has only 0 and 1. Only **480 frauds (0.192%)**.

## Observations
1. The data is clean, but **extremely imbalanced** (about 1 fraud per 520 transactions).
2. `hour_of_day`, `day_of_week`, `is_weekend` are copies of information inside `timestamp`.
3. `transaction id` is only a label, so it will not help a model.
4. Column names have spaces and brackets, so we rename them in cleaning.
5. Amounts go from Rs 10 to Rs 42,099, and most are small (median Rs 629).
